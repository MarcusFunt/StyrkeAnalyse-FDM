# Baseline tensile stress

Nominal tensile stress is the applied force divided by the specimen's cross-sectional area. With force in newtons and area in square millimetres, `1 N/mm² = 1 MPa`, so the result is directly in megapascals. Run the notebook from top to bottom after restarting the kernel.

```python
from pathlib import Path
import sys

sys.path.insert(0, str(Path.cwd() / "src"))
from fdm_strength.baseline import nominal_tensile_stress_mpa

force_newton = 850.0
area_mm2 = 40.0
stress_mpa = nominal_tensile_stress_mpa(force_newton, area_mm2)
print(f"Nominal tensile stress: {stress_mpa:.2f} MPa")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.bar(["Nominal tensile stress"], [stress_mpa])
ax.set_ylabel("Stress (MPa)")
ax.set_title("Baseline tensile stress")
plt.show()
```
